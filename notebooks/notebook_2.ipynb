```python
from prefect import task, flow
import duckdb
import pandas as pd
import sys
import os


# add parent directory to import path
current = os.getcwd()
parent = os.path.dirname(current)
sys.path.append(parent)

from src.features.feature_engineering import FeatureEngineer


@task
def extract(raw_path: str) -> pd.DataFrame:
    df = pd.read_csv(raw_path, sep=r'\s+', names=["unit number", "time, in cycles","operational setting 1","operational setting 2","operational setting 3","sensor measurement  1",
                                                                                            "sensor measurement  2","sensor measurement  3", "sensor measurement  4","sensor measurement  5",
                                                                                            "sensor measurement  6","sensor measurement  7","sensor measurement  8","sensor measurement  9",
                                                                                            "sensor measurement  10","sensor measurement  11", "sensor measurement  12","sensor measurement  13",
                                                                                            "sensor measurement  14","sensor measurement  15", "sensor measurement  16","sensor measurement  17",
                                                                                            "sensor measurement  18","sensor measurement  19", "sensor measurement  20","sensor measurement  21"])
    df["RUL"] = df.groupby("unit number")["time, in cycles"].transform("max") - df["time, in cycles"]
    df["RUL"] = df["RUL"].clip(upper=125)
    return df


@task
def transform(df: pd.DataFrame, window: int = 30) -> tuple[pd.DataFrame, FeatureEngineer]:
    fe = FeatureEngineer(window)
    fe.fit(df)
    smoothed_df = fe.transform(df)
    return smoothed_df, fe

@task
def load(df: pd.DataFrame, db_path: str, table_name: str) -> None:
    con = duckdb.connect(db_path)
    con.execute(f"DROP TABLE IF EXISTS {table_name}")
    con.execute(f"CREATE TABLE {table_name} AS SELECT * FROM df")

    con.close()

@flow
def etl_pipeline(raw_path: str, db_path: str, window: int = 30, table_name: str = "features_fd001") -> None:
    df = extract(raw_path)
    smoothed_df, fe = transform(df, window)
    load(smoothed_df, db_path, table_name)
    fe.save(parent + f"/artifacts/feature_engineer_{window}.joblib")
```

```python

etl_pipeline(
    raw_path=parent + "/data/raw/train_FD001.txt",
    db_path=parent + "/database/cmapss.db",
    window=30,
    table_name="features_30_fd001"
)
```

```python
import duckdb
import pandas as pd
import sys
import os


# add parent directory to import path
current = os.getcwd()
parent = os.path.dirname(current)
sys.path.append(parent)


con = duckdb.connect(parent + "/database/cmapss.db")


# query → pandas dataframe in one line
df = con.execute("SELECT * FROM features_fd001").df()
df
```

```python
import numpy as np

def make_sequences(df, window_length, feature_cols, target_col="RUL"):
    X_list = []
    y_list = []

    for _, group in df.groupby("unit number"):
        data = group[feature_cols].to_numpy()
        target = group[target_col].to_numpy()

        if len(group) < window_length:
            continue

        # sliding windows
        X = np.lib.stride_tricks.sliding_window_view(
            data, window_shape=(window_length, data.shape[1])
        ).squeeze(1)

        # targets aligned to last element in each window
        y = target[window_length - 1:]

        X_list.append(X)
        y_list.append(y)

    X = np.concatenate(X_list)
    y = np.concatenate(y_list)

    return X, y


feature_cols = [col for col in df.columns if col not in ["unit number", "time, in cycles", "RUL"]]

X, y = make_sequences(df, window_length=30, feature_cols=feature_cols)
```
